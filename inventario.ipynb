```python
#ingreso de la cantidad inicial de zapatos en inventario
inventario = int(input("ingrese la cantidad de pares de la tienda "))
transacciones = 0 #iniciamos el contador de transacciones
total_ventas = 0

#mientras haya zapatos en inventario
while inventario > 0:
  venta = int(input("¿Cuantos pares de zapatos se vendieron en esta transaccion? "))

  #deduce la venta del inventario y suma al contador de transacciones
  inventario -= venta
  transacciones +=1

  #verficiar que no se ingrese una venta mayor al inventario
  if inventario < 0:
    print("La venta excede el inventario actual, por favor ingrese numero valido")
    inventario += venta #devuelve la cantidad vendida al inventario
    transacciones -=1 #resta la transaccion invalida
  elif inventario >=0:
    total_ventas += venta
    print(f"Quedan {inventario} pares de zapatos en el inventario")

print(f"El inventario ha llegado a cero, se realizaron {transacciones} transacciones")
print(f"Se vendieron un total de {total_ventas} pares")
```
